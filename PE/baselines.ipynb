# Baselines on pursuit-evasion: R-PPO, DR-RPPO, VariBAD, PEARL, and MMD
Every baseline acts on the same observation interface as our pursuer, its own position, the evader signal, and its previous move $[p_k, y_k, a^1_{k-1}]$, through a recurrent policy. Every baseline trains until it has used the same number of environment steps as our method, read from `main_budget.json` (episodes end at capture, so steps rather than episodes are matched).

- **R-PPO**: recurrent PPO against the trained evader.
- **DR-RPPO**: recurrent PPO against an evader drawn per episode from a randomized pool.
- **VariBAD**: Bayes-adaptive meta-RL, conditioning the policy on an inferred task belief (mean and standard deviation of a recurrent variational encoder trained to predict the next observation).
- **PEARL**: latent-context meta-RL acting by posterior sampling, resampling the latent every `PEARL_RESAMPLE_K` steps from a product-of-Gaussians posterior over the episode's transitions.
- **MMD**: magnetic mirror descent in self-play, regularizing each player toward a slowly moving magnet copy of itself; it respects the game but never represents the context.

`RecurrentAC`, `VariBADEncoder`, `PEARLEncoder`, and `sample_config` are defined in `Initialize.ipynb`. Checkpoints are saved in `compared/`, each as a dictionary holding the network weights, the configuration, and the per-update training returns.

## Evader pool

```python
class OpponentTrainedP2:
    """The trained evader; it needs its private state m2."""
    needs_m2 = True
    def reset(self, Epi_Layout, exit_rc, Type_Theta):
        self.tt = torch.tensor(Type_Theta, dtype=torch.float32, device=device)
    def act(self, x, m2, prc, erc):
        with torch.no_grad():
            p = Player2_fixed(torch.tensor(x, dtype=torch.float32, device=device),
                              torch.tensor(m2, dtype=torch.float32, device=device),
                              self.tt).cpu().numpy()
        p = np.clip(p, 0, None); p = p / p.sum()
        return int(np.random.choice(n_u, p=p))

class OpponentRandom:
    """Uniform over the four moves."""
    needs_m2 = False
    def reset(self, Epi_Layout, exit_rc, Type_Theta):
        pass
    def act(self, x, m2, prc, erc):
        return int(np.random.randint(n_u))

class OpponentGreedyExit:
    """Moves to minimize Manhattan distance to the exit, ties broken at random."""
    needs_m2 = False
    def reset(self, Epi_Layout, exit_rc, Type_Theta):
        self.exit_rc = exit_rc
    def act(self, x, m2, prc, erc):
        dists = []
        for u in range(n_u):
            nr, nc = step_pos(erc[0], erc[1], u)
            dists.append(abs(nr - self.exit_rc[0]) + abs(nc - self.exit_rc[1]))
        dists = np.array(dists)
        return int(np.random.choice(np.where(dists == dists.min())[0]))

class OpponentEvasive:
    """Moves to maximize Manhattan distance to the pursuer, ties broken at random."""
    needs_m2 = False
    def reset(self, Epi_Layout, exit_rc, Type_Theta):
        pass
    def act(self, x, m2, prc, erc):
        dists = []
        for u in range(n_u):
            nr, nc = step_pos(erc[0], erc[1], u)
            dists.append(abs(nr - prc[0]) + abs(nc - prc[1]))
        dists = np.array(dists)
        return int(np.random.choice(np.where(dists == dists.max())[0]))

class OpponentExitEvasive:
    """Heads for the exit but switches to evasion when the pursuer is within 2 cells."""
    needs_m2 = False
    def reset(self, Epi_Layout, exit_rc, Type_Theta):
        self.greedy = OpponentGreedyExit(); self.greedy.reset(Epi_Layout, exit_rc, Type_Theta)
        self.evade = OpponentEvasive(); self.evade.reset(Epi_Layout, exit_rc, Type_Theta)
    def act(self, x, m2, prc, erc):
        if abs(prc[0] - erc[0]) + abs(prc[1] - erc[1]) <= 2:
            return self.evade.act(x, m2, prc, erc)
        return self.greedy.act(x, m2, prc, erc)

POOL_FACTORIES = [OpponentRandom, OpponentGreedyExit, OpponentEvasive,
                  OpponentExitEvasive, OpponentTrainedP2]
```

## Rollouts and PPO (episodes are padded to 40 steps and masked)

```python
def gae(rews, vals):
    T = len(rews)
    adv = np.zeros(T, dtype=np.float32); last = 0.0
    for t in reversed(range(T)):
        nv = vals[t + 1] if t + 1 < T else 0.0    # capture and timeout are both terminal
        delta = rews[t] + GAMMA * nv - vals[t]
        last = delta + GAMMA * GAE_LAMBDA * last
        adv[t] = last
    return adv, adv + np.asarray(vals, dtype=np.float32)

def rollout_episode(policy, opponent, hook=None):
    Epi_Layout, exit_rc, Theta, Type_Theta, p_row, p_col, e_row, e_col = sample_config()
    opponent.reset(Epi_Layout, exit_rc, Type_Theta)
    m2 = init_m2(e_row, e_col)
    h = None
    y_cur = np.zeros(d); a_prev = np.zeros(n_a)
    if hook is not None:
        hook.reset()
    obs_l, ex_l, a_l, lp_l, v_l, r_l, y_l, av_l = [], [], [], [], [], [], [], []
    for t in range(max_steps):
        own = np.zeros(d); own[p_row * 6 + p_col] = 1.0
        obs = np.concatenate([own, y_cur, a_prev]).astype(np.float32)
        obs_t = torch.tensor(obs, device=device).view(1, 1, OBS_DIM)
        ex_np = hook.extra(obs) if hook is not None else None
        ex_t = None if ex_np is None else torch.tensor(
            np.asarray(ex_np, dtype=np.float32), device=device).view(1, 1, -1)
        with torch.no_grad():
            logits, val, h = policy(obs_t, ex_t, h)
        dist = Categorical(logits=logits.view(-1))
        a = int(dist.sample())
        lp = float(dist.log_prob(torch.tensor(a, device=device)))
        x = build_x(p_row, p_col, e_row, e_col)
        u = opponent.act(x, m2, (p_row, p_col), (e_row, e_col))
        new_p_row, new_p_col, new_e_row, new_e_col = next_state(
            p_row, p_col, e_row, e_col, a, u, Epi_Layout)
        Captured = ((new_p_row, new_p_col) == (new_e_row, new_e_col)) or \
                   ((p_row, p_col) == (new_e_row, new_e_col) and
                    (e_row, e_col) == (new_p_row, new_p_col))
        timeout = (t == max_steps - 1)
        done = Captured or timeout
        r = float(Reward(new_p_row, new_p_col, new_e_row, new_e_col, exit_rc, timeout, Captured))
        if opponent.needs_m2:
            m2 = m2_update(m2, e_row, e_col, u, new_e_row, new_e_col)
        y = Observation(new_e_row, new_e_col)
        a_onehot = np.zeros(n_a, dtype=np.float32); a_onehot[a] = 1.0
        if hook is not None:
            hook.observe(y_cur, a_onehot, y)
        obs_l.append(obs); a_l.append(a); lp_l.append(lp); v_l.append(val.item())
        r_l.append(r); y_l.append(y.astype(np.float32)); av_l.append(a_onehot)
        if ex_np is not None:
            ex_l.append(np.asarray(ex_np, dtype=np.float32))
        p_row, p_col, e_row, e_col = new_p_row, new_p_col, new_e_row, new_e_col
        y_cur = y; a_prev = a_onehot
        if done:
            break
    adv, ret = gae(np.array(r_l, dtype=np.float32), np.array(v_l, dtype=np.float32))
    ep = {'obs': np.stack(obs_l), 'acts': np.array(a_l), 'logp': np.array(lp_l, dtype=np.float32),
          'adv': adv, 'ret': ret, 'rews': np.array(r_l, dtype=np.float32),
          'ys': np.stack(y_l), 'avecs': np.stack(av_l), 'len': len(a_l)}
    if ex_l:
        ep['extras'] = np.stack(ex_l)
    return ep

def stack_batch(eps, keys):
    E = len(eps)
    out = []
    for k in keys:
        dt = torch.long if k == 'acts' else torch.float32
        sample = np.asarray(eps[0][k])
        arr = np.zeros((E, max_steps) + sample.shape[1:],
                       dtype=np.int64 if k == 'acts' else np.float32)
        for i, e in enumerate(eps):
            arr[i, :e['len']] = e[k]
        out.append(torch.tensor(arr, dtype=dt, device=device))
    return out

def valid_mask(eps):
    m = torch.zeros(len(eps), max_steps, dtype=torch.float32, device=device)
    for i, e in enumerate(eps):
        m[i, :e['len']] = 1.0
    return m

def ppo_update(policy, opt, eps, extras=None, extra_params=None):
    obs, acts, oldlp, adv, ret = stack_batch(eps, ['obs', 'acts', 'logp', 'adv', 'ret'])
    mask = valid_mask(eps)
    mv = mask > 0
    adv = (adv - adv[mv].mean()) / (adv[mv].std() + 1e-8)
    params = list(policy.parameters()) + (list(extra_params) if extra_params else [])
    for _ in range(PPO_EPOCHS):
        logits, values, _ = policy(obs, extras)
        dist = Categorical(logits=logits)
        lp = dist.log_prob(acts)
        ratio = (lp - oldlp).exp()
        pg_t = -torch.min(ratio * adv,
                          torch.clamp(ratio, 1 - CLIP_EPS, 1 + CLIP_EPS) * adv)
        pg = (pg_t * mask).sum() / mask.sum()
        vl = (((values - ret) ** 2) * mask).sum() / mask.sum()
        ent = (dist.entropy() * mask).sum() / mask.sum()
        loss = pg + VF_COEF * vl - ENT_COEF * ent
        opt.zero_grad(); loss.backward()
        torch.nn.utils.clip_grad_norm_(params, 1.0)
        opt.step()
    return float(pg.item()), float(vl.item())
```

```python
def train_recurrent_ppo(save_name, opponent_sampler, seed):
    np.random.seed(seed); torch.manual_seed(seed); random.seed(seed)
    policy = RecurrentAC(extra_dim=0).to(device)
    opt = optim.AdamW(policy.parameters(), lr=LR)
    steps, upd = 0, -1
    ret_hist = []
    pbar = tqdm(total=BUDGET_STEPS, desc=save_name)
    while steps < BUDGET_STEPS:
        upd += 1
        eps = [rollout_episode(policy, opponent_sampler())
               for _ in range(EPISODES_PER_UPD)]
        n_new = sum(e['len'] for e in eps)
        steps += n_new; pbar.update(n_new)
        ppo_update(policy, opt, eps)
        ret_hist.append(float(np.mean([e['rews'].sum() for e in eps])))
        if (upd + 1) % CKPT_EVERY == 0 or steps >= BUDGET_STEPS:
            torch.save({'policy': policy.state_dict(),
                        'config': {'RNN_H': RNN_H, 'extra_dim': 0},
                        'return_per_episode': ret_hist},
                       os.path.join(SAVE_DIR, save_name + '.pth'))
        if (upd + 1) % 25 == 0:
            print(f"{save_name} upd {upd+1}  return/episode (last 25 upd): "
                  f"{np.mean(ret_hist[-25:]):.3f}")
    pbar.close()
    BUDGET[save_name] = steps
    return policy, ret_hist
```

## VariBAD: policy conditioned on a variational task belief

```python
class VariBADDecoder(nn.Module):
    """Predicts y_{t+1} from (z_t, y_t, a_t). Reward decoding dropped by design."""
    def __init__(self):
        super().__init__()
        self.net = nn.Sequential(nn.Linear(LATENT_DIM + d + n_a, 128), nn.ReLU(),
                                 nn.Linear(128, d))
    def forward(self, z, y, a):
        return self.net(torch.cat([z, y, a], dim=-1))

class VariBADHook:
    def __init__(self, enc):
        self.enc = enc
    def reset(self):
        self.h = None
    def extra(self, obs):
        with torch.no_grad():
            mu, logvar, self.h = self.enc(
                torch.tensor(obs, dtype=torch.float32, device=device).view(1, 1, -1), self.h)
        mu = mu.view(-1).cpu().numpy()
        std = np.exp(0.5 * logvar.view(-1).cpu().numpy())
        return np.concatenate([mu, std])
    def observe(self, y, a, y_next):
        pass

def train_varibad(seed=303):
    np.random.seed(seed); torch.manual_seed(seed); random.seed(seed)
    enc = VariBADEncoder().to(device)
    dec = VariBADDecoder().to(device)
    policy = RecurrentAC(extra_dim=2 * LATENT_DIM).to(device)
    enc_opt = optim.AdamW(list(enc.parameters()) + list(dec.parameters()), lr=LR)
    pol_opt = optim.AdamW(policy.parameters(), lr=LR)
    hook = VariBADHook(enc)
    steps, upd = 0, -1
    ret_hist = []
    pbar = tqdm(total=BUDGET_STEPS, desc='VARIBAD')
    while steps < BUDGET_STEPS:
        upd += 1
        eps = [rollout_episode(policy, OpponentTrainedP2(), hook)
               for _ in range(EPISODES_PER_UPD)]
        n_new = sum(e['len'] for e in eps)
        steps += n_new; pbar.update(n_new)
        # ---- ELBO step (encoder + decoder) ----
        obs, ys, avecs = stack_batch(eps, ['obs', 'ys', 'avecs'])
        mask = valid_mask(eps)
        mu, logvar, _ = enc(obs)
        std = (0.5 * logvar).exp()
        z = mu + std * torch.randn_like(std)
        recon = dec(z, obs[..., d:2 * d], avecs)
        rec_t = ((recon - ys) ** 2).mean(-1)
        kl_t = -0.5 * (1 + logvar - mu.pow(2) - logvar.exp()).sum(-1)
        elbo = (rec_t * mask).sum() / mask.sum() + VARIBAD_KL_W * (kl_t * mask).sum() / mask.sum()
        enc_opt.zero_grad(); elbo.backward()
        torch.nn.utils.clip_grad_norm_(list(enc.parameters()) + list(dec.parameters()), 1.0)
        enc_opt.step()
        # ---- PPO step; beliefs fixed to their acting-time values ----
        extras = stack_batch(eps, ['extras'])[0]
        ppo_update(policy, pol_opt, eps, extras=extras)
        ret_hist.append(float(np.mean([e['rews'].sum() for e in eps])))
        if (upd + 1) % CKPT_EVERY == 0 or steps >= BUDGET_STEPS:
            torch.save({'policy': policy.state_dict(),
                        'encoder': enc.state_dict(),
                        'decoder': dec.state_dict(),
                        'config': {'RNN_H': RNN_H, 'LATENT_DIM': LATENT_DIM,
                                   'extra_dim': 2 * LATENT_DIM},
                        'return_per_episode': ret_hist},
                       os.path.join(SAVE_DIR, 'VARIBAD.pth'))
        if (upd + 1) % 25 == 0:
            print(f"VARIBAD upd {upd+1}  return/episode (last 25 upd): "
                  f"{np.mean(ret_hist[-25:]):.3f}  elbo {elbo.item():.4f}")
    pbar.close()
    BUDGET['VARIBAD'] = steps
    return policy, enc, dec, ret_hist

VB_policy, VB_enc, VB_dec, VB_hist = train_varibad()
```

## PEARL: posterior sampling over a latent context

```python
class PEARLHook:
    def __init__(self, enc):
        self.enc = enc
    def reset(self):
        self.ctx = []
        self.z = np.random.randn(LATENT_DIM).astype(np.float32)   # prior sample
    def extra(self, obs):
        return self.z
    def observe(self, y, a, y_next):
        self.ctx.append(np.concatenate([y, a, y_next]).astype(np.float32))
        if len(self.ctx) % PEARL_RESAMPLE_K == 0:
            with torch.no_grad():
                mu, var = self.enc.posterior(
                    torch.tensor(np.stack(self.ctx), device=device))
            self.z = (mu + var.sqrt() * torch.randn_like(mu)).cpu().numpy()

def train_pearl(seed=404):
    np.random.seed(seed); torch.manual_seed(seed); random.seed(seed)
    enc = PEARLEncoder().to(device)
    policy = RecurrentAC(extra_dim=LATENT_DIM).to(device)
    opt = optim.AdamW(list(policy.parameters()) + list(enc.parameters()), lr=LR)
    hook = PEARLHook(enc)
    steps, upd = 0, -1
    ret_hist = []
    pbar = tqdm(total=BUDGET_STEPS, desc='PEARL')
    while steps < BUDGET_STEPS:
        upd += 1
        eps = [rollout_episode(policy, OpponentTrainedP2(), hook)
               for _ in range(EPISODES_PER_UPD)]
        n_new = sum(e['len'] for e in eps)
        steps += n_new; pbar.update(n_new)
        obs, acts, oldlp, adv, ret = stack_batch(eps, ['obs', 'acts', 'logp', 'adv', 'ret'])
        mask = valid_mask(eps)
        mv = mask > 0
        adv = (adv - adv[mv].mean()) / (adv[mv].std() + 1e-8)
        extras_act = stack_batch(eps, ['extras'])[0]
        trans_list = [torch.tensor(
            np.concatenate([e['obs'][:, d:2 * d], e['avecs'], e['ys']], axis=-1),
            dtype=torch.float32, device=device) for e in eps]
        E = len(eps)
        for _ in range(PPO_EPOCHS):
            # actor pass: acting-time latents, detached
            logits, _, _ = policy(obs, extras_act)
            dist = Categorical(logits=logits)
            lp = dist.log_prob(acts)
            ratio = (lp - oldlp).exp()
            pg_t = -torch.min(ratio * adv,
                              torch.clamp(ratio, 1 - CLIP_EPS, 1 + CLIP_EPS) * adv)
            pg = (pg_t * mask).sum() / mask.sum()
            ent = (dist.entropy() * mask).sum() / mask.sum()
            # critic pass: full-context posterior sample, gradient trains the encoder
            mus, vars_, zs = [], [], []
            for e_i in range(E):
                mu, var = enc.posterior(trans_list[e_i])
                zs.append(mu + var.sqrt() * torch.randn_like(mu))
                mus.append(mu); vars_.append(var)
            z_seq = torch.stack(zs).unsqueeze(1).expand(-1, max_steps, -1)
            _, values, _ = policy(obs, z_seq)
            vl = (((values - ret) ** 2) * mask).sum() / mask.sum()
            mu_s = torch.stack(mus); var_s = torch.stack(vars_)
            kl = 0.5 * (var_s + mu_s.pow(2) - 1.0 - var_s.log()).sum(-1).mean()
            loss = pg + VF_COEF * vl - ENT_COEF * ent + PEARL_KL_W * kl
            opt.zero_grad(); loss.backward()
            torch.nn.utils.clip_grad_norm_(
                list(policy.parameters()) + list(enc.parameters()), 1.0)
            opt.step()
        ret_hist.append(float(np.mean([e['rews'].sum() for e in eps])))
        if (upd + 1) % CKPT_EVERY == 0 or steps >= BUDGET_STEPS:
            torch.save({'policy': policy.state_dict(),
                        'encoder': enc.state_dict(),
                        'config': {'RNN_H': RNN_H, 'LATENT_DIM': LATENT_DIM,
                                   'extra_dim': LATENT_DIM,
                                   'resample_k': PEARL_RESAMPLE_K},
                        'return_per_episode': ret_hist},
                       os.path.join(SAVE_DIR, 'PEARL.pth'))
        if (upd + 1) % 25 == 0:
            print(f"PEARL upd {upd+1}  return/episode (last 25 upd): "
                  f"{np.mean(ret_hist[-25:]):.3f}")
    pbar.close()
    BUDGET['PEARL'] = steps
    return policy, enc, ret_hist

PEARL_policy, PEARL_enc, PEARL_hist = train_pearl()
```

## MMD: magnetic mirror descent in self-play

```python
class P2AC(nn.Module):
    def __init__(self):
        super().__init__()
        self.trunk = nn.Sequential(nn.Linear(3 * d + 4, 256), nn.ReLU(),
                                   nn.Linear(256, 256), nn.ReLU())
        self.pi = nn.Linear(256, n_u)
        self.v = nn.Linear(256, 1)
    def forward(self, inp):
        h = self.trunk(inp)
        return self.pi(h), self.v(h).squeeze(-1)

def rollout_selfplay(p1, p2):
    Epi_Layout, exit_rc, Theta, Type_Theta, p_row, p_col, e_row, e_col = sample_config()
    tt = Type_Theta.astype(np.float32)
    m2 = init_m2(e_row, e_col)
    h = None
    y_cur = np.zeros(d); a_prev = np.zeros(n_a)
    o1, a1, l1, v1, r1 = [], [], [], [], []
    i2, u2, l2, v2 = [], [], [], []
    for t in range(max_steps):
        own = np.zeros(d); own[p_row * 6 + p_col] = 1.0
        obs = np.concatenate([own, y_cur, a_prev]).astype(np.float32)
        with torch.no_grad():
            logits, val, h = p1(torch.tensor(obs, device=device).view(1, 1, OBS_DIM), None, h)
        d1 = Categorical(logits=logits.view(-1))
        a = int(d1.sample())
        x = build_x(p_row, p_col, e_row, e_col)
        inp2 = np.concatenate([x, m2, tt]).astype(np.float32)
        with torch.no_grad():
            logits2, val2 = p2(torch.tensor(inp2, device=device).unsqueeze(0))
        d2 = Categorical(logits=logits2.view(-1))
        u = int(d2.sample())
        new_p_row, new_p_col, new_e_row, new_e_col = next_state(
            p_row, p_col, e_row, e_col, a, u, Epi_Layout)
        Captured = ((new_p_row, new_p_col) == (new_e_row, new_e_col)) or \
                   ((p_row, p_col) == (new_e_row, new_e_col) and
                    (e_row, e_col) == (new_p_row, new_p_col))
        timeout = (t == max_steps - 1)
        done = Captured or timeout
        r = float(Reward(new_p_row, new_p_col, new_e_row, new_e_col, exit_rc, timeout, Captured))
        m2n = m2_update(m2, e_row, e_col, u, new_e_row, new_e_col)
        y = Observation(new_e_row, new_e_col)
        a_onehot = np.zeros(n_a, dtype=np.float32); a_onehot[a] = 1.0
        o1.append(obs); a1.append(a)
        l1.append(float(d1.log_prob(torch.tensor(a, device=device))))
        v1.append(val.item()); r1.append(r)
        i2.append(inp2); u2.append(u)
        l2.append(float(d2.log_prob(torch.tensor(u, device=device))))
        v2.append(val2.item())
        p_row, p_col, e_row, e_col = new_p_row, new_p_col, new_e_row, new_e_col
        y_cur = y; a_prev = a_onehot; m2 = m2n
        if done:
            break
    r1 = np.array(r1, dtype=np.float32)
    adv1, ret1 = gae(r1, np.array(v1, dtype=np.float32))
    adv2, ret2 = gae(-r1, np.array(v2, dtype=np.float32))
    L = len(a1)
    ep1 = {'obs': np.stack(o1), 'acts': np.array(a1),
           'logp': np.array(l1, dtype=np.float32), 'adv': adv1, 'ret': ret1,
           'rews': r1, 'len': L}
    ep2 = {'inp': np.stack(i2), 'acts': np.array(u2),
           'logp': np.array(l2, dtype=np.float32), 'adv': adv2, 'ret': ret2, 'len': L}
    return ep1, ep2

def mmd_update_p1(p1, mag1, opt, eps):
    obs, acts, oldlp, adv, ret = stack_batch(eps, ['obs', 'acts', 'logp', 'adv', 'ret'])
    mask = valid_mask(eps)
    mv = mask > 0
    adv = (adv - adv[mv].mean()) / (adv[mv].std() + 1e-8)
    for _ in range(PPO_EPOCHS):
        logits, values, _ = p1(obs)
        dist = Categorical(logits=logits)
        with torch.no_grad():
            mag_logits, _, _ = mag1(obs)
        kl_t = kl_divergence(dist, Categorical(logits=mag_logits))
        kl_mag = (kl_t * mask).sum() / mask.sum()
        lp = dist.log_prob(acts)
        ratio = (lp - oldlp).exp()
        pg_t = -torch.min(ratio * adv,
                          torch.clamp(ratio, 1 - CLIP_EPS, 1 + CLIP_EPS) * adv)
        pg = (pg_t * mask).sum() / mask.sum()
        vl = (((values - ret) ** 2) * mask).sum() / mask.sum()
        ent = (dist.entropy() * mask).sum() / mask.sum()
        loss = pg + VF_COEF * vl - ENT_COEF * ent + MMD_COEF * kl_mag
        opt.zero_grad(); loss.backward()
        torch.nn.utils.clip_grad_norm_(p1.parameters(), 1.0)
        opt.step()

def mmd_update_p2(p2, mag2, opt, eps2):
    inp = torch.tensor(np.concatenate([e['inp'] for e in eps2]),
                       dtype=torch.float32, device=device)
    acts = torch.tensor(np.concatenate([e['acts'] for e in eps2]),
                        dtype=torch.long, device=device)
    oldlp = torch.tensor(np.concatenate([e['logp'] for e in eps2]),
                         dtype=torch.float32, device=device)
    adv = torch.tensor(np.concatenate([e['adv'] for e in eps2]),
                       dtype=torch.float32, device=device)
    ret = torch.tensor(np.concatenate([e['ret'] for e in eps2]),
                       dtype=torch.float32, device=device)
    adv = (adv - adv.mean()) / (adv.std() + 1e-8)
    for _ in range(PPO_EPOCHS):
        logits, values = p2(inp)
        dist = Categorical(logits=logits)
        with torch.no_grad():
            mag_logits, _ = mag2(inp)
        kl_mag = kl_divergence(dist, Categorical(logits=mag_logits)).mean()
        lp = dist.log_prob(acts)
        ratio = (lp - oldlp).exp()
        pg = -torch.min(ratio * adv,
                        torch.clamp(ratio, 1 - CLIP_EPS, 1 + CLIP_EPS) * adv).mean()
        vl = F.mse_loss(values, ret)
        loss = pg + VF_COEF * vl - ENT_COEF * dist.entropy().mean() + MMD_COEF * kl_mag
        opt.zero_grad(); loss.backward()
        torch.nn.utils.clip_grad_norm_(p2.parameters(), 1.0)
        opt.step()

def ema_(magnet, net):
    with torch.no_grad():
        for m, c in zip(magnet.parameters(), net.parameters()):
            m.mul_(1.0 - MAGNET_TAU).add_(MAGNET_TAU * c)

def train_mmd(seed=505):
    np.random.seed(seed); torch.manual_seed(seed); random.seed(seed)
    p1 = RecurrentAC(extra_dim=0).to(device)
    p2 = P2AC().to(device)
    mag1 = deepcopy(p1); mag2 = deepcopy(p2)
    mag1.gru.flatten_parameters()
    for m in list(mag1.parameters()) + list(mag2.parameters()):
        m.requires_grad_(False)
    opt1 = optim.AdamW(p1.parameters(), lr=LR)
    opt2 = optim.AdamW(p2.parameters(), lr=LR)
    steps, upd = 0, -1
    ret_hist = []
    pbar = tqdm(total=BUDGET_STEPS, desc='MMD')
    while steps < BUDGET_STEPS:
        upd += 1
        pairs = [rollout_selfplay(p1, p2)
                 for _ in range(EPISODES_PER_UPD)]
        n_new = sum(e1['len'] for e1, _ in pairs)
        steps += n_new; pbar.update(n_new)
        eps1 = [a for a, b in pairs]; eps2 = [b for a, b in pairs]
        mmd_update_p1(p1, mag1, opt1, eps1)
        mmd_update_p2(p2, mag2, opt2, eps2)
        ema_(mag1, p1); ema_(mag2, p2)
        ret_hist.append(float(np.mean([e['rews'].sum() for e in eps1])))
        if (upd + 1) % CKPT_EVERY == 0 or steps >= BUDGET_STEPS:
            torch.save({'p1': p1.state_dict(), 'p2': p2.state_dict(),
                        'p1_magnet': mag1.state_dict(), 'p2_magnet': mag2.state_dict(),
                        'config': {'RNN_H': RNN_H, 'extra_dim': 0,
                                   'MMD_COEF': MMD_COEF, 'MAGNET_TAU': MAGNET_TAU},
                        'return_per_episode': ret_hist},
                       os.path.join(SAVE_DIR, 'MMD.pth'))
        if (upd + 1) % 25 == 0:
            print(f"MMD upd {upd+1}  P1 return/episode (last 25 upd): "
                  f"{np.mean(ret_hist[-25:]):.3f}")
    pbar.close()
    BUDGET['MMD'] = steps
    return p1, p2, ret_hist

MMD_p1, MMD_p2, MMD_hist = train_mmd()
```
